## Part 1: Setup and expense entry

Begin with the user setup, then allow the user to enter as many expenses as they like in one session.

The budget validation below is provided as a worked pattern. Your task is the expense entry loop beneath it. Keep asking for expenses until the user types `done`, and for each valid amount, store it, add it to the running total, and give brief feedback.


```python
3
# Budget validation, provided as a worked pattern.
# Keeps asking until the number falls within a sensible range.
monthly_budget = 0
while monthly_budget < 100 or monthly_budget > 10000:
    try:
        budget_input = input("Enter monthly budget ($100 to $10,000): $")
        monthly_budget = float(budget_input)
        if monthly_budget < 100:
            print("Budget too low. Minimum is $100.")
        elif monthly_budget > 10000:
            print("Budget seems high. Maximum is $10,000.")
    except ValueError:
        print("Please enter a valid number.")

print(f"\nWelcome {name}. Monthly budget: ${monthly_budget:.2f}")

# ===== Your task: multi-expense entry =====
expenses = []      # every amount is stored here
total_spent = 0

print("\n--- Enter your expenses ---")
print("Type 'done' when you are finished.")

while True:
    entry = input("\nExpense amount (or 'done'): $")

    if entry.lower() == "done":
        break

    # Your task:
    # 1. try to convert entry into a float
    # 2. if it is not a positive number, tell the user and continue
    # 3. if it is valid, append it to expenses, add it to total_spent,
    #    and print the amount along with the running total
    try:
        expense_amount = float(entry)
        if expense_amount <= 0:
            print("Please enter a positive amount.")
            continue
        expenses.append(expense_amount)
        total_spent += expense_amount
        print(f"  Added: ${expense_amount:.2f} | Running total: ${total_spent:.2f}")
    except ValueError:
        print("Please enter a valid number.")


print(f"\nEntry complete: {len(expenses)} expenses, ${total_spent:.2f} total")

```

## Part 2: Summarise the numbers

You now have a list of expenses, so the next step is to summarise it. The total and average are provided. Your task is to find the highest and lowest values, and to count the large expenses.


```python

if len(expenses) > 0:
    print("\n--- Analysis ---")
    average = total_spent / len(expenses)
    print(f"Total spent: ${total_spent:.2f}")
    print(f"Average expense: ${average:.2f}")
    print(f"Budget used: {(total_spent / monthly_budget) * 100:.1f}%")

    # Your task: find the highest and lowest expense.
    # Start both at the first item, then loop and compare.
    highest = expenses[0]
    lowest = expenses[0]

    for expense in expenses:
        if expense > highest:
            highest = expense
        if expense < lowest:
            lowest = expense

    print(f"Highest expense: ${highest:.2f}")
    print(f"Lowest expense: ${lowest:.2f}")

    # Your task: count how many expenses were more than 20% of the budget.
    large_count = 0
    threshold = monthly_budget * 0.20

    for expense in expenses:
        if expense > threshold:
            large_count += 1

    print(f"Large expenses (over 20% of budget): {large_count}")

else:
    print("\nNo expenses entered, so there is nothing to analyse.")
```

## Part 3: A menu that keeps running

Wrap the program in a menu so the user can keep working until they choose to quit. Option 4 (quit) is provided. Complete options 1, 2 and 3 using patterns from earlier in the notebook.


```python
# Finance Tracker Program with Functions

# Variables are initialized in previous cells (174582c3, 2bfa4cab)
# name
# expenses
# total_spent
# monthly_budget


# ===== FUNCTIONS =====

def display_menu():
    """Display the menu options"""
    print("\n" + "-" * 40)
    # 'name' is expected to be a global variable from previous cell
    print(f"{name}'s finance tracker")
    print("-" * 40)
    print("1. View all expenses")
    print("2. View summary")
    print("3. Add an expense")
    print("4. Delete an expense") # Added delete option
    print("5. Quit")
    print("-" * 40)


def view_expenses():
    """Option 1: View all expenses"""
    # 'expenses' and 'total_spent' are expected to be global variables
    global expenses, total_spent

    if not expenses:
        print("\nNo expenses recorded yet.")
    else:
        print("\nYour expenses:")
        for i, expense in enumerate(expenses, 1):
            print(f"  {i}. ${expense:.2f}")
        print(f"\nTotal: ${total_spent:.2f}")


def view_summary():
    """Option 2: View summary"""
    # 'expenses', 'total_spent', 'monthly_budget' are expected to be global variables
    global expenses, total_spent, monthly_budget

    if not expenses:
        print("\nNo expenses recorded yet.")
        print(f"Total spent: $0.00")
        print(f"Number of expenses: 0")
    else:
        average = total_spent / len(expenses)
        highest = max(expenses)
        lowest = min(expenses)

        print(f"\n--- Summary ---")
        print(f"Total spent: ${total_spent:.2f}")
        print(f"Number of expenses: {len(expenses)}")
        print(f"Average expense: ${average:.2f}")
        print(f"Highest expense: ${highest:.2f}")
        print(f"Lowest expense: ${lowest:.2f}")

        # Budget analysis from previous cell
        print(f"Budget used: {(total_spent / monthly_budget) * 100:.1f}%")

        large_count = 0
        threshold = monthly_budget * 0.20
        for expense in expenses:
            if expense > threshold:
                large_count += 1
        print(f"Large expenses (over {threshold:.2f}): {large_count}")


def add_expense():
    """Option 3: Add an expense"""
    # 'expenses' and 'total_spent' are expected to be global variables
    global expenses, total_spent

    try:
        expense_amount = float(input("\nEnter expense amount: $"))
        if expense_amount > 0:
            expenses.append(expense_amount)
            total_spent += expense_amount
            print(f"Expense of ${expense_amount:.2f} added successfully!")
            print(f"Current total: ${total_spent:.2f}")
        else:
            print("Please enter a positive amount.")
    except ValueError:
        print("Invalid input. Please enter a valid number.")


def delete_expense():
    """Option 4: Delete an expense"""
    global expenses, total_spent # Declare globals

    if not expenses:
        print("\nNo expenses to delete.")
        return

    print("\nYour expenses:")
    for i, expense in enumerate(expenses, 1):
        print(f"  {i}. ${expense:.2f}")

    try:
        choice = int(input("\nEnter the number of the expense to delete: "))

        if 1 <= choice <= len(expenses):
            deleted = expenses.pop(choice - 1)
            total_spent -= deleted
            print(f"Deleted: ${deleted:.2f}")
            print(f"New total: ${total_spent:.2f}")
        else:
            print("Invalid choice.")

    except ValueError:
        print("Please enter a valid number.")


def main():
    """Main program loop"""
    global name, total_spent # Declare name and total_spent as global for final message

    running = True

    while running:
        display_menu()
        choice = input("\nChoose an option (1-5): ") # Options changed to 1-5

        if choice == "1":
            view_expenses()
        elif choice == "2":
            view_summary()
        elif choice == "3":
            add_expense()
        elif choice == "4":
            delete_expense()
        elif choice == "5": # Quit option changed to 5
            print(f"\nGoodbye, {name}! Final total: ${total_spent:.2f}") # Using total_spent
            running = False
        else:
            print("Invalid choice. Please choose 1-5.")


# ===== RUN PROGRAM =====
if __name__ == "__main__":
    main()
```
