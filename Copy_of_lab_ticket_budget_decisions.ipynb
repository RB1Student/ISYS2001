## Your final prompt

Using the Python Programming language create a tool that is a saving goal tracker that givers feedback depending on progress of the saving goal.
The input Should include
- The Saving Goal Name and Amount
- Variable Cost per Week Name and Amount that is paid the same each (an option if you want to change or keep that weeks cost the same
- Cost for the week they came up (random expenses or cost) - Current Saving amount and can store the amount in the systems
- How much they have earned that week

Output
- Tell the Consumer the % to goal
- How long it might take to reach the goal
- If they are going negative for the week or positive for the week in savings

```


````python
import json
import os
import math

DATA_FILE = "saving_goal.json"


class SavingGoalTracker:
    def __init__(self):
        self.data = self.load_data()

    # -------------------------
    # Load / Save Data
    # -------------------------
    def load_data(self):
        if os.path.exists(DATA_FILE):
            with open(DATA_FILE, "r") as file:
                return json.load(file)
        return {}

    def save_data(self):
        with open(DATA_FILE, "w") as file:
            json.dump(self.data, file, indent=4)

    # -------------------------
    # Create Goal
    # -------------------------
    def create_goal(self):
        print("\n=== Create Saving Goal ===")

        goal_name = input("Saving Goal Name: ")
        goal_amount = float(input("Goal Amount ($): "))
        current_savings = float(input("Current Savings ($): "))

        fixed_name = input("Weekly Fixed Cost Name: ")
        fixed_amount = float(input("Weekly Fixed Cost Amount ($): "))

        self.data = {
            "goal_name": goal_name,
            "goal_amount": goal_amount,
            "current_savings": current_savings,
            "fixed_cost_name": fixed_name,
            "fixed_cost_amount": fixed_amount
        }

        self.save_data()
        print("\nSaving goal created successfully!\n")

    # -------------------------
    # Weekly Update
    # -------------------------
    def weekly_update(self):

        if not self.data:
            print("No saving goal found.")
            return

        print("\n=== Weekly Update ===")

        print(
            f"Current Fixed Cost: "
            f"{self.data['fixed_cost_name']} "
            f"(${self.data['fixed_cost_amount']:.2f})"
        )

        keep = input("Keep this cost? (Y/N): ").lower()

        if keep == "n":
            self.data["fixed_cost_name"] = input("New Fixed Cost Name: ")
            self.data["fixed_cost_amount"] = float(
                input("New Fixed Cost Amount ($): ")
            )

        income = float(input("Income earned this week ($): "))
        random_expenses = float(input("Random expenses this week ($): "))

        fixed_cost = self.data["fixed_cost_amount"]

        weekly_balance = income - fixed_cost - random_expenses

        self.data["current_savings"] += weekly_balance

        if self.data["current_savings"] < 0:
            self.data["current_savings"] = 0

        self.save_data()

        self.show_progress(weekly_balance)

    # -------------------------
    # Display Results
    # -------------------------
    def show_progress(self, weekly_balance):

        goal = self.data["goal_amount"]
        savings = self.data["current_savings"]

        percent = (savings / goal) * 100

        print("\n========== RESULT ==========")

        print(f"Goal: {self.data['goal_name']}")
        print(f"Goal Amount: ${goal:.2f}")
        print(f"Current Savings: ${savings:.2f}")
        print(f"Progress: {percent:.2f}%")

        if weekly_balance > 0:
            print(f"\nYou SAVED ${weekly_balance:.2f} this week.")
        elif weekly_balance < 0:
            print(f"\nYou LOST ${abs(weekly_balance):.2f} this week.")
        else:
            print("\nYou broke even this week.")

        if weekly_balance > 0:

            remaining = goal - savings

            if remaining <= 0:
                print("\n🎉 Congratulations!")
                print("You have reached your saving goal!")
            else:
                weeks = math.ceil(remaining / weekly_balance)
                print(f"Estimated weeks remaining: {weeks}")

        else:
            print(
                "At your current spending rate, "
                "you are not moving toward your goal."
            )

        print("=============================\n")

    # -------------------------
    # View Goal
    # -------------------------
    def view_goal(self):

        if not self.data:
            print("No saving goal has been created.")
            return

        print("\n===== Current Goal =====")
        print(f"Goal Name: {self.data['goal_name']}")
        print(f"Goal Amount: ${self.data['goal_amount']:.2f}")
        print(f"Current Savings: ${self.data['current_savings']:.2f}")
        print(
            f"Fixed Weekly Cost: "
            f"{self.data['fixed_cost_name']} "
            f"(${self.data['fixed_cost_amount']:.2f})"
        )
        print("=========================\n")


# -------------------------
# Main Menu
# -------------------------
tracker = SavingGoalTracker()

while True:

    print("===== Saving Goal Tracker =====")
    print("1. Create Saving Goal")
    print("2. Weekly Update")
    print("3. View Goal")
    print("4. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        tracker.create_goal()

    elif choice == "2":
        tracker.weekly_update()

    elif choice == "3":
        tracker.view_goal()

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice.\n")
````
