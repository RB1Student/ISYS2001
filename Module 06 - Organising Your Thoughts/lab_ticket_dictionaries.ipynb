## The three functions your code has to get right

Build your dictionary however you like, then write these three functions to work on it.
The checks at the bottom call them by name, so use these names and have them behave as
described:

- `total_for(spending, category)` returns the amount spent in that category, and returns
  `0` for a category that was never recorded. It must not crash on a missing category.
- `add_transaction(spending, transaction)` takes one transaction, in the same shape as
  the records in the list above, and updates your dictionary so that amount is counted
  from then on.
- `summary(spending)` returns a dictionary that maps every category to its total.

How `spending` is built and shaped inside is entirely your decision. The checks only care
about what these three functions do, not how you store things underneath.

```python
#1


spending = {}
for transaction_item in transactions:
    category = transaction_item['category']
    if category: # Only add if category is not blank
        amount = transaction_item['amount']
        spending[category] = spending.get(category, 0) + amount

def total_for(spending, category):
    """Returns the amount spent in that category, and returns 0 for a category that was never recorded."""
    return spending.get(category, 0)

def add_transaction(spending, transaction):
    """Adds a new transaction to the spending dictionary. If a transaction has no 'category' key, its amount is not added to any category."""
    category = transaction.get('category')
    if category: # Only add if category is not blank
        amount = transaction['amount']
        spending[category] = spending.get(category, 0) + amount

def summary(spending):
    """Returns a dictionary that maps every category to its total."""
    return spending

# You can test it out here
print("Initial Spending Summary:")
for category in spending:
    print(f"Total for {category}: {total_for(spending, category)}")
print(f"Total for made up category: {total_for(spending, 'made up category')}")
```

```python
#2

# Get the current total for rent
current_rent = total_for(spending, 'rent')

# Calculate the amount to add to reach 1500
amount_to_add = 1500 - current_rent

# Create a transaction to adjust the rent
rent_adjustment_transaction = {'date': '2026-08-31', 'category': 'rent', 'amount': amount_to_add, 'description': 'Rent adjustment'}

# Add the transaction using the add_transaction function
add_transaction(spending, rent_adjustment_transaction)

# Print the updated total for rent
print(f"Updated total for rent: {total_for(spending, 'rent')}")

```

```python
#3

all_expenses_summary = summary(spending)
total_expenses = sum(all_expenses_summary.values())
print(f"Total of all expenses: {total_expenses}")
```
